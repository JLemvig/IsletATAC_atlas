```python
import os

data_dir = os.path.abspath("data/public")
project_dir = os.path.abspath("data/project")
chrombpnet_dir = os.path.join(project_dir, "chrombpnet")

import pandas as pd
```

```python
def setup_peaks(peak_file, size = 500):
    peak_file = peak_file.copy()
    center = ((peak_file[1] + peak_file[2]) // 2)
    peak_file[1] = center - (size // 2)
    peak_file[2] = center + (size // 2)
    peak_file[1] = peak_file[1].astype(int)
    peak_file[2] = peak_file[2].astype(int)
    peak_file[3] = peak_file[0] + ":" + peak_file[1].astype(str) + "-" + peak_file[2].astype(str)
    peak_file[5] = "."
    peak_file[6] = peak_file[4]
    peak_file[7] = 0
    peak_file[8] = 0
    peak_file[9] = size // 2
    peak_file = peak_file.drop_duplicates(subset = 3)
    peak_file = peak_file.sort_values(by=[0, 1])
    peak_file = peak_file.reset_index(drop=True)
    return peak_file

## Prep beta peaks
# Read
pc4 = pd.read_csv(f"{data_dir}/scATACseq/correlations/beta/pc1pc2pc3pc5pc6pc7pc18/DESeq_normalized_nogc/bed/pc4.bed", sep="\t", header=None)
pc9 = pd.read_csv(f"{data_dir}/scATACseq/correlations/beta/pc1pc2pc3pc5pc6pc7pc18/DESeq_normalized_nogc/bed/pc9.bed", sep="\t", header=None)
pc10 = pd.read_csv(f"{data_dir}/scATACseq/correlations/beta/pc1pc2pc3pc5pc6pc7pc18/DESeq_normalized_nogc/bed/pc10.bed", sep="\t", header=None)
pc11 = pd.read_csv(f"{data_dir}/scATACseq/correlations/beta/pc1pc2pc3pc5pc6pc7pc18/DESeq_normalized_nogc/bed/pc11.bed", sep="\t", header=None)
pc13 = pd.read_csv(f"{data_dir}/scATACseq/correlations/beta/pc1pc2pc3pc5pc6pc7pc18/DESeq_normalized_nogc/bed/pc13.bed", sep="\t", header=None)

# Combine
beta_peaks = pd.concat((pc4, pc9, pc10, pc11, pc13))

# Setup
beta_peaks = setup_peaks(beta_peaks)
beta_peaks.to_csv(f"{chrombpnet_dir}/data/peaks/beta_correlated.bed", sep="\t", index=False, header=False)

## Prep alpha peaks
# Read
pc4 = pd.read_csv(f"{data_dir}/scATACseq/correlations/alpha/pc1pc2pc3pc5pc6pc7pc18/DESeq_normalized_nogc/bed/pc4.bed", sep="\t", header=None)
pc9 = pd.read_csv(f"{data_dir}/scATACseq/correlations/alpha/pc1pc2pc3pc5pc6pc7pc18/DESeq_normalized_nogc/bed/pc9.bed", sep="\t", header=None)
pc10 = pd.read_csv(f"{data_dir}/scATACseq/correlations/alpha/pc1pc2pc3pc5pc6pc7pc18/DESeq_normalized_nogc/bed/pc10.bed", sep="\t", header=None)
pc11 = pd.read_csv(f"{data_dir}/scATACseq/correlations/alpha/pc1pc2pc3pc5pc6pc7pc18/DESeq_normalized_nogc/bed/pc11.bed", sep="\t", header=None)
pc13 = pd.read_csv(f"{data_dir}/scATACseq/correlations/alpha/pc1pc2pc3pc5pc6pc7pc18/DESeq_normalized_nogc/bed/pc13.bed", sep="\t", header=None)

# Combine
alpha_peaks = pd.concat((pc4, pc9, pc10, pc11, pc13))

# Setup
alpha_peaks = setup_peaks(alpha_peaks)
alpha_peaks.to_csv(f"{chrombpnet_dir}/data/peaks/alpha_correlated.bed", sep="\t", index=False, header=False)
```
